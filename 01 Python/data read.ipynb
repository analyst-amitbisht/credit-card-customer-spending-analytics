```python
import pandas as pd

customers = pd.read_csv(r"dim_customers.csv")
spends = pd.read_csv(r"fact_spends.csv")
```

## **dim_customers**

### **1. About**

```python
customers.sample()
```

```python
customers.info()
```

### **2. Summary**

```python
customers.describe().round(2)
```

```python
customers.describe(include = ['object', str])
```

```python
profile = pd.DataFrame({
    "unique": customers.nunique(),
    "uniqueness": (customers.nunique() / len(customers)).round(4),
    "missing_values": customers.isna().sum(),
    "missing_value_pct": (customers.isna().mean()*100).round(2),
    "duplicates": customers.duplicated().sum(),
})
print(profile.sort_values("uniqueness", ascending=False))
```

```python
print(f"{customers['age_group'].value_counts(dropna=False)}\n")
print(f"{customers['city'].value_counts(dropna=False)}\n")
print(f"{customers['occupation'].value_counts(dropna=False)}\n")
print(f"{customers['gender'].value_counts(dropna=False)}\n")
print(f"{customers['marital status'].value_counts(dropna=False)}\n")
```

### **3. Primary Key Validation**

```python
print("Customer IDs unique:",
      customers["customer_id"].is_unique)

print("Customer IDs missing:",
      customers["customer_id"].isna().sum())
```

### **4. Foreign Key Validation**

```python
invalid_booking_key = customers[~customers['customer_id'].isin(spends['customer_id'])]
print(f"Invalid booking_id records: {len(invalid_booking_key)}")
```

### **5. Whitespace Validation**

```python
categorical_cols = customers.select_dtypes(include=['str', 'object']).columns
for col in categorical_cols:
    whitespace_rows = customers[customers[col] != customers[col].str.strip()]
    print(f"{col}: {len(whitespace_rows)} rows with leading/trailing whitespace")
```
